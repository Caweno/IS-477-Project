```python
def get_quarter_label(month):
    if pd.isna(month):
        return None
    
    if month in ['January','February', 'March']:
        return f"January-March 2024"
    elif month in ['April','May','June']:
        return f"April-June 2024"
    elif month in ['July','August','September']:
        return f"July-September 2024"
    else:
        return f"October-December 2024"

df_all["Quarter"] = df_all["Month"].apply(get_quarter_label)
```
